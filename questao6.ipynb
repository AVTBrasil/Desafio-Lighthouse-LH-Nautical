```python
import pandas as pd 
import matplotlib as pyplot
import numpy as np
from sklearn.metrics import mean_absolute_error
```

Carregando os 4 arquivos

```python
produtos = pd.read_csv('./1-lh_nautical_csv/products.csv')
variantes_produtos = pd.read_csv('./1-lh_nautical_csv/product_variants.csv')
pedidos = pd.read_csv('./1-lh_nautical_csv/orders.csv')
itens_pedido = pd.read_csv('./1-lh_nautical_csv/order_items.csv')
```

Unificar Datasets

```python
# primeiro preciso achar as colunas comum entre eles
print(produtos.columns)
print(variantes_produtos.columns)
print(pedidos.columns)
print(itens_pedido.columns)
```

```python
arq_1 = pd.merge(
    produtos,
    variantes_produtos,
    left_on='id',
    right_on='product_id',
    suffixes=('_produto', '_variante')
)

arq_2 = pd.merge(
    arq_1,
    itens_pedido,
    left_on='id_variante',
    right_on='product_variant_id'
)

arq_final = pd.merge(
    arq_2,
    pedidos,
    left_on='order_id',
    right_on='id'
)
```

```python
arq_final.head()
```

Filtrar "Bússola de Bordo 702"

```python
bussola = arq_final[arq_final['name'] == "Bússola de Bordo 702"].copy()
bussola.head()
```

```python
print(bussola.shape)
```

```python
# conferindo as datas 

bussola['placed_at'].min(), bussola['placed_at'].max()
```

```python
# como a datas no futuro vamos tratar

bussola = bussola[
    bussola['placed_at'] <= '2026-03-31'
].copy()
```

Criar vendas mensais

```python
print(bussola['placed_at'].dtype) 
# tem que ser data
```

```python
bussola['placed_at'] = pd.to_datetime(bussola['placed_at'])
```

```python
vendas_mensais = (
    bussola.groupby(pd.Grouper(key = 'placed_at', freq = 'MS'))['quantity']
    .sum().reset_index()
)

vendas_mensais
```

```python
# verificando vendas mensais

print(vendas_mensais['placed_at'].min())
print(vendas_mensais['placed_at'].max())
print(len(vendas_mensais['placed_at']))
```

Separar treino e teste 

```python
treino = vendas_mensais[
    vendas_mensais['placed_at'] <= '2025-12-31'
].copy()
teste = vendas_mensais[
    (vendas_mensais['placed_at'] >= '2026-01-01') &
    (vendas_mensais['placed_at'] < '2026-03-31')
]
```

```python
print("Treino: ")
print(treino)
print("Teste: ")
print(teste)
```

Calcular média movel em três meses

```python
treino['previsao'] = (
    treino['quantity'].rolling(window = 3).mean().shift(1)
)

treino.head(10)
```

```python
# rolling pega os três meses e shift impede que a quantidade do proprio mes entre na previsão
vendas_mensais['media_3'] =(
    vendas_mensais['quantity'].rolling(window = 3).mean().shift(1) 
)
```

```python
treino.tail()
```

Gerar Previsões para Jan/Fev/Mar de 2026

```python
teste_previsao = vendas_mensais[
    (vendas_mensais['placed_at'] >= '2026-01-01') &
    (vendas_mensais['placed_at'] < '2026-03-31')    
].copy()
print(teste_previsao)
```

Calcular MAE 'Erro quadrático médio'

```python
mae = mean_absolute_error(teste_previsao['quantity'], teste_previsao['media_3'])
print(mae)
```

ou seja, aproximadamente 19 unidades por mês
